```python
def find_name(text):
    for line in text.split("\n"):
        line = line.strip()
        if line != "" and line.lower() not in ["resume", "curriculum vitae", "cv"]:
            return line
    return "Not found"


def find_email(text):
   
    allowed = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789._-+"

    position = text.find("@")                      
    while position != -1:
        start = position                          
        while start > 0 and text[start - 1] in allowed:
            start = start - 1

        end = position + 1                        
        while end < len(text) and text[end] in allowed:
            end = end + 1

        email = text[start:end].strip(".-_+")
       
        if start < position and "." in text[position:end]:
            return email

        position = text.find("@", position + 1)  
    return "Not found"


def find_phone(text):
    digits = ""
    for ch in text + "\n":
        if ch.isdigit():
            digits = digits + ch
        elif ch != "\n" and (ch.isspace() or ch in "-()+."):
            pass                                  
        else:
           
            if len(digits) >= 10 and len(digits) <= 13 and digits[-10] in "6789":
                return digits[-10:]
            digits = ""
    return "Not found"
```

```python
MARKS = [",", ";", ":", "(", ")", "[", "]", "|", "/", "&", "\n", "\t",
         "\u2022", "\u25cf", "\u2013", "\u2014", "'", "\u2019"]

DEGREES = ["b.tech", "m.tech", "btech", "mtech", "b.e", "m.e", "bca", "mca", "b.sc", "m.sc", "bsc", "msc",
           "mba", "bba", "b.com", "m.com", "bachelor", "bachelors", "master", "masters", "diploma", "phd"]


def load_skills(file_name="skills.txt"):
    skills = []
    with open(file_name) as f:
        for line in f:
            skill = line.strip().lower()
            if skill != "":
                skills.append(skill)
    return skills


def clean_text(text):
    text = " " + text.lower() + " "
    for mark in MARKS:
        text = text.replace(mark, " ")
    text = text.replace(". ", " ")
    return text


def find_skills(text, skills):
    text = clean_text(text)
    found = []
    for skill in skills:
        if " " + skill + " " in text:
            found.append(skill)
    return found


def find_education(text):
    result = []
    for line in text.split("\n"):
        line_clean = clean_text(line)
        for degree in DEGREES:
            if " " + degree + " " in line_clean:
                result.append(line.strip())
                break
    return result


def parse_resume(text, skills):
    return {
        "name": find_name(text),
        "email": find_email(text),
        "phone": find_phone(text),
        "skills": find_skills(text, skills),
        "education": find_education(text),
    }
```

```python
def match_skills(resume_skills, job_skills):
    matched = []
    missing = []
    for skill in job_skills:
        if skill in resume_skills:
            matched.append(skill)
        else:
            missing.append(skill)

    if len(job_skills) == 0:
        score = 0
    else:
        score = round(len(matched) / len(job_skills) * 100)

    return {"score": score, "matched": matched, "missing": missing}


def get_job_match(resume_skills, jd_text):
    job_skills = find_skills(jd_text, load_skills())
    return match_skills(resume_skills, job_skills)
```
