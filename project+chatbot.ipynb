```python
def find_name(text):
    for line in text.split("\n"):
        line = line.strip()
        if line != "" and line.lower() not in ["resume", "curriculum vitae", "cv"]:
            return line
    return "Not found"


def find_email(text):
   
    allowed = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789._-+"

    position = text.find("@")                      
    while position != -1:
        start = position                          
        while start > 0 and text[start - 1] in allowed:
            start = start - 1

        end = position + 1                        
        while end < len(text) and text[end] in allowed:
            end = end + 1

        email = text[start:end].strip(".-_+")
       
        if start < position and "." in text[position:end]:
            return email

        position = text.find("@", position + 1)  
    return "Not found"


def find_phone(text):
    digits = ""
    for ch in text + "\n":
        if ch.isdigit():
            digits = digits + ch
        elif ch != "\n" and (ch.isspace() or ch in "-()+."):
            pass                                  
        else:
           
            if len(digits) >= 10 and len(digits) <= 13 and digits[-10] in "6789":
                return digits[-10:]
            digits = ""
    return "Not found"
```

## function to fetch details  from resume

```python
MARKS = [",", ";", ":", "(", ")", "[", "]", "|", "/", "&", "\n", "\t",
         "\u2022", "\u25cf", "\u2013", "\u2014", "'", "\u2019"]

DEGREES = ["b.tech", "m.tech", "btech", "mtech", "b.e", "m.e", "bca", "mca", "b.sc", "m.sc", "bsc", "msc",
           "mba", "bba", "b.com", "m.com", "bachelor", "bachelors", "master", "masters", "diploma", "phd"]


def load_skills(file_name="skills.txt"):
    skills = []
    with open(file_name) as f:
        for line in f:
            skill = line.strip().lower()
            if skill != "":
                skills.append(skill)
    return skills


def clean_text(text):
    text = " " + text.lower() + " "
    for mark in MARKS:
        text = text.replace(mark, " ")
    text = text.replace(". ", " ")
    return text


def find_skills(text, skills):
    text = clean_text(text)
    found = []
    for skill in skills:
        if " " + skill + " " in text:
            found.append(skill)
    return found


def find_education(text):
    result = []
    for line in text.split("\n"):
        line_clean = clean_text(line)
        for degree in DEGREES:
            if " " + degree + " " in line_clean:
                result.append(line.strip())
                break
    return result


def parse_resume(text, skills):
    return {
        "name": find_name(text),
        "email": find_email(text),
        "phone": find_phone(text),
        "skills": find_skills(text, skills),
        "education": find_education(text),
    }
```

## the chatbot prompt and function

```python
def match_skills(resume_skills, job_skills):
    matched = []
    missing = []
    for skill in job_skills:
        if skill in resume_skills:
            matched.append(skill)
        else:
            missing.append(skill)

    if len(job_skills) == 0:
        score = 0
    else:
        score = round(len(matched) / len(job_skills) * 100)

    return {"score": score, "matched": matched, "missing": missing}


def get_job_match(resume_skills, jd_text):
    job_skills = find_skills(jd_text, load_skills())
    return match_skills(resume_skills, job_skills)


def make_prompt(parsed, match, jd_text):
    return f"""You are a friendly career coach for the candidate below.
Use the resume, the job description and the match result given below.

Rules:
- Start with one line that explains the score: how many of the job skills (from our skills list) the resume has.
- Then list the missing skills and, for each one, say what to do: what to learn first, how to practice it with a small project, and how to show it on the resume.
- Do not guess future scores. Do not invent course names, links or company names. Suggest types of free resources (official documentation, YouTube tutorials, practice sites) and small project ideas.
- The job description may mention skills that are not in our lists. Those are not counted in the score. Say so if they matter.
- Only talk about this resume and this job. If the question is about something else, politely say you can only help with the resume and the job.
- Do not invent facts about the candidate.
- Keep the answer under 180 words. Format with Markdown: a bold title for each missing skill, and short bullet points below it.

RESUME
Name: {parsed['name']}
Email: {parsed['email']}
Phone: {parsed['phone']}
Education: {', '.join(parsed['education'])}
Skills in resume: {', '.join(parsed['skills'])}

JOB DESCRIPTION
{jd_text}

MATCH RESULT
Score: {match['score']}%
Matched skills: {', '.join(match['matched'])}
Missing skills: {', '.join(match['missing'])}
"""


def ask_bot(question, parsed, match, jd_text, history):
    try:
        messages = [{"role": "system", "content": make_prompt(parsed, match, jd_text)}]
        for old_message in history[-6:]:
            messages.append(old_message)
        messages.append({"role": "user", "content": question})

        response = client.chat.completions.create(model=MODEL, messages=messages)
        answer = response.choices[0].message.content

        history.append({"role": "user", "content": question})        # save only if it worked
        history.append({"role": "assistant", "content": answer})
    except Exception as error:
        answer = "Sorry, something went wrong. The real error is: " + str(error)
    return answer
```

```python
from IPython.display import display, Markdown


def chat(parsed):
    history = []
    match = None
    jd_text = ""
    print("Paste the job description and press Enter. Type exit to stop.")

    while True:
        if match is None:
            jd_text = input("Job description: ")
            if jd_text.strip().lower() == "exit":
                break

            match = get_job_match(parsed["skills"], jd_text)
            if len(match["matched"]) + len(match["missing"]) == 0:
                print("No known skills found in this job description. Paste a longer description or add skills to skills.txt.")
                match = None
                continue

            display(Markdown(
                "### Match score: " + str(match["score"]) + " %\n\n"
                + "**Matched skills:** " + (", ".join(match["matched"]) or "none") + "\n\n"
                + "**Missing skills:** " + (", ".join(match["missing"]) or "none")
            ))
            first_question = "What is my match score, which skills are missing, and what should I do to improve?"
            answer = ask_bot(first_question, parsed, match, jd_text, history)
            display(Markdown("### Bot\n\n" + answer))
            print("Ask more questions. Type new job for another job description, or exit to stop.")
        else:
            question = input("You: ")
            if question.strip().lower() == "exit":
                break
            if question.strip().lower() == "new job":
                match = None
                history = []
                print("Paste the new job description.")
                continue
            answer = ask_bot(question, parsed, match, jd_text, history)
            display(Markdown("**Bot:**\n\n" + answer))

    print("Bye!")
```

```python
import gradio as gr

FIRST_QUESTION = "What is my match score, which skills are missing, and what should I do to improve?"


def start_chat(file_name):
    file_name = file_name.strip().strip('"')
    if file_name == "":
        return [], {}, "Please paste the resume file name first."

    try:
        resume_text = read_pdf(file_name)
    except Exception as error:
        return [], {}, "Could not open the file: " + str(error)

    if resume_text == "":
        return [], {}, "No text found in this PDF. It may be a scanned image."

    parsed = parse_resume(resume_text, load_skills())

    details = (
        "### Resume details\n"
        "- **Name:** " + parsed["name"] + "\n"
        "- **Email:** " + parsed["email"] + "\n"
        "- **Phone:** " + parsed["phone"] + "\n"
        "- **Education:** " + (", ".join(parsed["education"]) or "Not found") + "\n"
        "- **Skills:** " + (", ".join(parsed["skills"]) or "none found") + "\n\n"
        "Now please **paste the job description** below and press Enter."
    )
    chat_history = [{"role": "assistant", "content": details}]
    state = {"parsed": parsed, "match": None, "jd_text": "", "history": []}
    return chat_history, state, "Resume loaded."


def send_message(message, chat_history, state):
    message = message.strip()
    if message == "":
        return "", chat_history, state
    if "parsed" not in state:
        return "", chat_history, state

    chat_history = chat_history + [{"role": "user", "content": message}]

    if message.lower() == "new job":
        state["match"] = None
        state["history"] = []
        reply = "Okay. Please paste the new job description."

    elif state["match"] is None:
        match = get_job_match(state["parsed"]["skills"], message)
        if len(match["matched"]) + len(match["missing"]) == 0:
            reply = "I could not find any known skills in this job description. Please paste a longer description, or add skills to skills.txt."
        else:
            state["match"] = match
            state["jd_text"] = message
            answer = ask_bot(FIRST_QUESTION, state["parsed"], match, message, state["history"])
            reply = (
                "### Match score: " + str(match["score"]) + " %\n"
                "- **Matched skills:** " + (", ".join(match["matched"]) or "none") + "\n"
                "- **Missing skills:** " + (", ".join(match["missing"]) or "none") + "\n\n"
                + answer + "\n\n"
                "*Ask more questions, or type `new job` to give another job description.*"
            )

    else:
        reply = ask_bot(message, state["parsed"], state["match"], state["jd_text"], state["history"])

    chat_history = chat_history + [{"role": "assistant", "content": reply}]
    return "", chat_history, state


try:
    demo.close()
except NameError:
    pass

with gr.Blocks(title="Resume Analyzer") as demo:
    gr.Markdown("# Resume Analyzer with Chat Assistant\n"
                "1. Paste the resume file name and click **Start**.  2. Paste the job description.  3. Ask your questions.")

    with gr.Row():
        file_box = gr.Textbox(label="Resume file name", placeholder="example: my_resume.pdf", scale=4)
        start_button = gr.Button("Start", variant="primary", scale=1)
    status = gr.Markdown("")

    try:
        chatbot = gr.Chatbot(type="messages", height=480)
    except TypeError:
        chatbot = gr.Chatbot(height=480)

    message_box = gr.Textbox(label="Your message", placeholder="Paste the job description or type a question, then press Enter")
    state = gr.State({})

    start_button.click(start_chat, [file_box], [chatbot, state, status])
    file_box.submit(start_chat, [file_box], [chatbot, state, status])
    message_box.submit(send_message, [message_box, chatbot, state], [message_box, chatbot, state])

demo.launch()
```
